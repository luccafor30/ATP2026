```python
import random

entrada = input("Você deseja entrar no menu do programa?")

while entrada != "sim" and entrada != "Sim" and entrada != "Não" and entrada != "não" and entrada != "nao" and entrada != "Nao":
    print("Não entendi, responda apenas sim ou não.")
    entrada = input("Você deseja entrar no menu do programa?")
if entrada == "nao" or entrada == "não":
    print("Entendido, a aplicação foi encerrada. Até à próxima!")
elif entrada == "Sim" or entrada == "sim":
    opcao = "-1"
    while opcao != "0":
        print("     Aplicação para gestão de alunos      ")
        print("(1) Criar uma turma")
        print("(2) Inserir um aluno na turma")
        print("(3) Listar a turma")
        print("(4) Consultar um aluno por id ")
        print("(5) Guardar a turma em ficheiro")
        print("(6) Carregar uma turma dum ficheiro")
        print("(0) Sair")
        opcao = input("Escolha uma opção").strip()

        if opcao == "1":
            print("Escolheu criar uma turma, agora adicione os alunos seguindo a opção 2")
            turma=[]

        elif opcao=="2":
            print("Escolheu inserir um aluno na turma")
            nomeAluno = input("Qual é o nome do aluno?").strip()
            idAluno = input("Qual é o ID do aluno?(ex:A123456)")
            notaTpc = random.randint(0,20)
            notaProj = random.randint(0,20)
            notaTeste = random.randint(0,20)
            notasAluno = [notaTpc, notaProj, notaTeste]
            pacoteAluno = (nomeAluno, idAluno, notasAluno)
            turma.append(pacoteAluno)
            print(f"O aluno {nomeAluno} foi adicionado à turma com os seus dados.")


        elif opcao=="3":
            print("\n--- Lista da Turma ---")
            if len(turma) > 0:
                for aluno in turma:
                    nome = aluno[0]
                    idAluno = aluno[1]
                    notasAluno = aluno[2]
                    print(f"ID: {idAluno} | Nome: {nome} | Notas (TPC, Projeto, Teste): {notasAluno}")
                print("-------------------\n")
            else:
                print("A turma ainda está vazia. Insira alunos na Opção 2 primeiro.")

        elif opcao=="4":
            print("\n--- Consultar Aluno por ID ---")
            if len(turma) > 0:
                procurarID = input("Digite o ID do aluno que deseja procurar (ex: A123456): ").strip()
                encontrei=False
                i = 0
                while i < len(turma) and encontrei==False:
                    if turma[i][1] == procurarID:
                        nome = turma[i][0]
                        notasAluno = turma[i][2]
                        print(f"Aluno encontrado. Nome: {nome} | ID: {procurarID} | Notas: {notasAluno}")
                        encontrei = True
                    i = i + 1
                if encontrei==False:
                    print(f"Não foi encontrado nenhum aluno com o ID {procurarID}")
            else:
                print("A turma ainda está vazia. Insira alunos primeiro (Opção 2).")
            print("------------------------------\n")

        elif opcao == '5':
            print("\n--- Guardar a turma em ficheiro ---")
            if len(turma) > 0:
                nomeFicheiro = input("Qual o nome do ficheiro para guardar (ex: turma.txt)? ").strip()
                file = open(nomeFicheiro, "w")
                for aluno in turma:
                    nome, id_aluno, notas = aluno
                    file.write(f"{nome},{id_aluno},{notas}\n")
                file.close()
                print(f"Os dados foram guardados no ficheiro {nomeFicheiro}")
            else:
                print("A turma ainda está vazia. Não há dados para guardar.")


        elif opcao == "6":
            print("\n--- Carregar uma turma dum ficheiro ---")
            if len(turma) > 0:
                print("Aviso, carregar um ficheiro vai substituir a turma atual.")
            nomeFicheiro = input("Qual o nome do ficheiro a carregar (ex: turma.txt)? ").strip()
            turma=[]
            file=open(nomeFicheiro, "r")
            for linha in file:
                linhaLimpa = linha.strip()
                if len(linhaLimpa) > 0:
                    partes = linhaLimpa.strip(",")
                    nome = partes[0]
                    idAluno= partes[1]
                    notasTexto = partes[2] + "," + partes[3] + "," + partes[4]
                    alunoRecuperado= (nome, idAluno, notasTexto)
                    turma.append(alunoRecuperado)
            file.close()
            print(f"A turma foi carregada. Estão agora {len(turma)} alunos ativos.")
            print("----------------------------------\n")

        elif opcao=="0":
            print(f"Encerrando a aplicação.")


```
