```python
#PARTE 1


#a)
lista1 = [1, 2, 3, 4, 5]                                                    
lista2 = [4, 5, 6, 7, 8]                                                    
ncomuns = [x for x in lista1 + lista2 if (x in lista1) != (x in lista2)]    
print(ncomuns)                                                              

#b)
texto = """Vivia há já não poucos anos algures num concelho do Ribatejo 
    um pequeno lavrador e negociante de gado chamado Manuel Peres Vigário"""
lista = [palavra for palavra in texto.split() if len(palavra) > 3]   

#c)
lista = ['anaconda', 'burro', 'cavalo', 'macaco']
listaRes = []

i = 1
for animal in lista:
    listaRes.append((i, animal))
    i = i + 1
print (listaRes)

#PARTE 2


#a)
def strCount(s, subs):
    contador = 0  
    tamanho = len(subs) 
    i = 0 

    while i <= (len(s) - tamanho):    
        if s[i : i + tamanho] == subs:      
            contador = contador + 1        
            i = i + tamanho                 
        else: 
            i = i + 1                       

    return contador

print(strCount("catcowcat", "cat")) 
print(strCount("catcowcat", "cow")) 
print(strCount("catcowcat", "dog")) 

#b)
def produtoM3(lista):
    lista.sort()  
    resultado = lista[0] * lista[1] * lista[2] 
    return  resultado

print(produtoM3([12,3,7,10,12,8,9]))

#c)
def reduxInt(n):
    textoNum = str(n)  
    while len(textoNum) > 1:    
        soma = 0
        for digito in textoNum:    
            soma = soma + int(digito)       
        textoNum = str(soma)               
    return int(textoNum)                   

print(reduxInt(38))  
print(reduxInt(777)) 

#d)
def myIndexOf(s1, s2):
    tamanho = len(s2)
    i = 0
    while i < len(s1) - tamanho:
        if s1[i : i + tamanho] == s2:
            return i 
        else:
            i = i + 1
    return -1

print(myIndexOf("Hoje está um belo dia de sol!", "belo"))
print(myIndexOf("Hoje está um belo dia de sol!", "chuva"))

#PARTE 3


#a)
def quantosPost(redeSocial):
    totalPosts = len(redeSocial) 
    return totalPosts

#b)
def postsAutor(redeSocial, autor):
    listaAutor = [post for post in redeSocial if post.get('autor')== autor]
    return listaAutor

#c)
def autores(redeSocial):
    todosAutores = [post.get('autor') for post in redeSocial if post.get('autor') is not None] 
    autoresUnicos = list(set(todosAutores))         
    autoresUnicos.sort()
    return autoresUnicos

#d)
def insPost(redeSocial, conteudo, autor, dataCriacao, comentarios):
    novoNumero = len(redeSocial) + 1    
    novoId = 'p' + str(novoNumero)      
    
    novoPost = {        
        'id': novoId,
        'conteudo': conteudo,
        'autor': autor,
        'dataCriacao': dataCriacao,
        'comentarios': comentarios
    }

    redeSocial.append(novoPost) 
    
    return redeSocial

#e)
def remPost(redeSocial, id):
    redeSocial = [post for post in redeSocial if post.get('id') != id]
    return redeSocial

#f)
def postsPorAutor(redeSocial):
    distribuicao = {}
    
    for post in redeSocial:
        autor = post.get('autor') 
        if autor in distribuicao:
            distribuicao[autor] = distribuicao[autor] + 1
        else:
            distribuicao[autor] = 1
            
    return distribuicao

#g)
def comentadoPor(redeSocial, autor):
    postsEncontrados = [] 

    for post in redeSocial: 
        listaComentarios = post.get('comentarios', []) 

        for comentario in listaComentarios: 

            if comentario.get('autor') == autor: 

                if post not in postsEncontrados: 
                    postsEncontrados.append(post) 
                    
    return postsEncontrados 
```
