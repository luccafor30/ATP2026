### Aquecimento: Calcula quantos números inteiros entre 1 e 10000 são múltiplos de 23 e devolve esse número (um múltiplo de 23 é um número que tem 23 como divisor).

```
def multiplos23():
    ... 
    return ... 
```

````python
def multiplos23():
    res = 0
    for n in range(1,10001):
        if n % 23==0:
            res = res + 1
    return res

print(multiplos23())
print(10000/23)
````

````python
def multiplos23():
    res = 0
    for n in range(23,10001,23):
        res = res + 1
    return res

print(multiplos23())
print(10000/23)
````

### 1.c) Define uma função para carregar uma tabela meteorológica de um ficheiro de texto.

````python
def carregaTabMeteo(fnome):
    res = []
    file = open(fnome, "r")
    for linha in file:
        linhaLimpa = linha.strip()
        if len(linhaLimpa)>0:
            partes1 = linhaLimpa.split("|")
            partes2 = partes1[0].split("/")
            ano = partes2[0]
            mes = partes2[1]
            dia = partes2[2]
            tmin = partes1[1]
            tmax = partes1[2]
            prec = partes1[3]
            texto = ((int(ano), int(mes), int(dia)), float(tmin), float(tmax), float(prec))
            res.append(texto)
    file.close()
    return res

tabMeteo2 = carregaTabMeteo("meteorologia.txt")
print(tabMeteo2)
````
