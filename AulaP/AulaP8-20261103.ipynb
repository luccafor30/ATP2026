### P1: Fatorial

O fatorial de um número natural $n$ ($n!$) é o produto de todos os inteiros positivos menores ou iguais a $n$:
* **Caso Base**: $0! = 1$ e $1! = 1$
* **Passo Recursivo**: $n! = n \times (n - 1)!$, para $n > 1$

Especifica a função `fatorial(n)` de forma recursiva.

````python
def fatorial(n):
    """Calcula o fatorial de n recursivamente (n >= 0)."""
    if n<=1:
        return 1
    else:
        return n*fatorial(n-1)

print(fatorial(5))
````

````python
# Casos de teste para P1
print("0! =", fatorial(0))   # Esperado: 1
print("1! =", fatorial(1))   # Esperado: 1
print("5! =", fatorial(5))   # Esperado: 120
print("7! =", fatorial(7))   # Esperado: 5040
````

### P2: Sequência de Fibonacci

A sucessão de Fibonacci é definida recursivamente por:
* $fib(1) = 1$
* $fib(2) = 1$
* $fib(n) = fib(n - 1) + fib(n - 2)$, para $n > 2$

Especifica uma função recursiva `fib(n)` que devolve o $n$-ésimo número da sucessão de Fibonacci.

````python
def fib(n):
    """Devolve o n-ésimo termo da sequência de Fibonacci (n >= 1)."""
    if n<=2:
        return 1
    return fib(n-1)+fib(n-2)
````

### P3: Exponenciação (Potência)

Especifica uma função recursiva `pot(base, exp)` que calcula a potência $base^{exp}$ para inteiros ($exp \ge 0$):
* **Caso Base**: $base^0 = 1$
* **Passo Recursivo**: $base^{exp} = base \times base^{exp - 1}$

No final, implementa também a versão iterativa `potIter(base, exp)`.

````python
def pot(base, exp):
    """Calcula base ** exp recursivamente (exp >= 0)."""
    if exp == 0:
        return 1
    return base*pot(base, exp-1)

def potIter(base, exp):
    """Calcula base ** exp iterativamente (exp >= 0)."""
    res = 1
    while exp > 0:
        res = res * base
        exp = exp - 1

    return res
    
````

````python
# Casos de teste para P3
print("Recursivo 2^5:", pot(2, 5))         # Esperado: 32
print("Recursivo 3^0:", pot(3, 0))         # Esperado: 1
print("Iterativo 2^5:", potIter(2, 5))     # Esperado: 32
print("Iterativo 3^0:", potIter(3, 0))     # Esperado: 1
````

### P7: Inverter uma String

Especifica uma função recursiva `inverteString(s)` que inverte uma string sem recorrer a fatiamento inverso (`s[::-1]`).

* Exemplo: `inverteString("algoritmo")` $\to$ `"omtirogla"`.
* *Dica indutiva*: A inversão de uma string não-vazia é o seu último carácter concatenado com a inversão do resto.

````python
def inverteString(s):
    """Inverte recursivamente a string s."""
    if len(s)== 0:
        return ""
    else:
       return inverteString(s[1:])+ s[0]

    
````

### P9: Contagem de Ocorrências de um Elemento

Dada uma lista de elementos e um valor `alvo`, conta o número de vezes que `alvo` ocorre na lista utilizando uma função recursiva.

* Exemplo: Para `lista = [1, 2, 3, 2, 4, 2]` e `alvo = 2`, o resultado deve ser `3`.

````python
def contaOcorr(lista, alvo):
    """Conta recursivamente quantas vezes alvo ocorre em lista."""
    if lista == []:
        return 0
    else:
        if lista[0] == alvo:
            return contaOcorr(lista[1:], alvo) + 1
        else:
            return contaOcorr(lista[1:], alvo)
````

### P12: Pesquisa Binária

A **pesquisa binária** localiza eficientemente um elemento `x` numa lista previamente **ordenada**, dividindo sucessivamente o espaço de pesquisa a metade.

Dada uma lista ordenada e um valor `x`, a função deve retornar o **índice** desse valor ou `-1` caso não exista.

Implementa:
1. A versão recursiva `pesquisaBin(lista, x, inicio, fim)`;
2. A versão iterativa `pesquisaBinIter(lista, x)`.

````python
def pesquisaBin(lista, x, inicio=0, fim=None):
    """Executa pesquisa binária recursiva numa lista ordenada."""
    if fim is None:
        fim = len(lista) - 1
    if inicio > fim:
        return -1
    meio = (inicio + fim)//2
    if x == lista[meio]:
        return meio
    if x > lista[meio]:
       return pesquisaBin(lista, x, meio + 1, fim )
    else:
       return pesquisaBin(lista, x, inicio, meio - 1)





def pesquisaBinIter(lista, x):
    """Executa pesquisa binária iterativa numa lista ordenada."""
    inicio = 0
    fim = len(lista) - 1
    res = -1
    encontrei = False
    while not encontrei and inicio <= fim:
        meio= (inicio + fim)//2
        if lista[meio] == x:
            encontrei = True
            res = meio
        else:
            if x > lista[meio]:
                inicio = meio + 1
            else:
                fim = meio - 1

    return res

        
    
    

        

````

````python
# Casos de teste para P12
dados = [10, 23, 35, 42, 58, 64, 77, 89, 91]
print("Recursiva - Procura 42:", pesquisaBin(dados, 42))        # Esperado: 3
print("Recursiva - Procura 50:", pesquisaBin(dados, 50))        # Esperado: -1
print("Iterativa - Procura 42:", pesquisaBinIter(dados, 42))    # Esperado: 3
print("Iterativa - Procura 50:", pesquisaBinIter(dados, 50))    # Esperado: -1
````
