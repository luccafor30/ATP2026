#### a) Especifique uma função que dada uma string e uma substring não vazia, calcula  o número de vezes em que a substring aparece na string, sem que haja sobreposição de substrings:

````python
def strCount(s, subs):
    contador = 0  #é o contador q vair permitir contar quantas vezes ele vai encontrar uma palavra numa string, por isso começa em 0
    tamanho = len(subs) #guardamos o nº de letras q estamos a procurar
    i = 0 #´o indice de leitura, começa procurando desde a primeira letra do lado esquerdo, dps vai passando letra a letra no ciclo

    while i <= (len(s) - tamanho):    #é a barreira, o ciclo vai parar quando faltar ler menos letras do q aquelas q a palavra tem ( n vale a pena procurar pela palavra cat se tiver apenas uma letra no fim da string)
        if s[i : i + tamanho] == subs:      #  s[ondeComeça : ondeTermina], no primeiro começa na posição 0 e vai 0 + 3 da primeira palavra cat, daí surge s[0:3], daí ele corta da palavra grande as posições 1,2,3, formando a palavra cat. Se o texto q cortamos for igual a palavra q perocuramos dá certo 
            contador = contador + 1         # dando certo o contador incrementa essa palavra lá q é os resultados q vamos ter
            i = i + tamanho                 # aqui fazemos isto para ele pular direto essa palavra tpda q encontramos e passar ela a frente pra evitar conta desnecessária
        else: 
            i = i + 1                       # aqui, no caso de falhar, passamos a letra seguinte pra se analisar no novo inicio do ciclo

    return contador

print(strCount("catcowcat", "cat")) # --> 2
print(strCount("catcowcat", "cow")) # --> 1
print(strCount("catcowcat", "dog")) # --> 0
````

#### c) Especifique uma função que dado um número inteiro positivo, repetidamente adiciona os seus dígitos até obter apenas um dígito que é retornado como resultado:

````python
# Input: 38
# Output: 2
# Explicação: 3 + 8 = 11, 1 + 1 = 2.

# Input: 777
# Output: 3
# Explicação: 7 + 7 + 7 = 21, 2 + 1 = 3.

def reduxInt(n):
    textoNum = str(n)  # Passamos o número original para texto logo de início
    while len(textoNum) > 1:    # O ciclo repete-se enquanto o texto tiver MAIS de 1 letra (dígito)
        soma = 0
        for digito in textoNum:    # lemos cada letra (dígito) da string
            soma = soma + int(digito)       # Passamos de volta a inteiro e somamos à caixa
        textoNum = str(soma)               # Quando acaba de somar, transformamos o novo resultado em string de novo, para voltar ao topo do ciclo while e ver se já é só 1 dígito
    return int(textoNum)                   # Quando o while termina (len <= 1), passamos para inteiro pela última vez para devolver

print(reduxInt(38))  
print(reduxInt(777)) 
````

#### d) Especifique uma função que recebe duas strings, `string1` e `string2`, e devolve o índice da primeira ocorrência de `string2` em `string1`, caso não ocorra nenhuma vez a função deverá retornar `-1`:

````python
# Invocação: indexOf("Hoje está um belo dia de sol!", "belo")
# Resultado: 13

# Invocação: indexOf("Hoje está um belo dia de sol!", "chuva")
# Resultado: -1

def myIndexOf(s1, s2):
    tamanho = len(s2)
    i = 0
    while i < len(s1) - tamanho:
        if s1[i : i + tamanho] == s2:
            return i 
        else:
            i = i + 1
    return -1

print(myIndexOf("Hoje está um belo dia de sol!", "belo"))
print(myIndexOf("Hoje está um belo dia de sol!", "chuva"))
````

Defina as seguintes funções de manipulação e consulta da rede social:

#### b)  `postsAutor`, que devolve a lista de posts de um determinado autor:

#### c) `autores`, que devolve a lista de autores de posts ordenada alfabeticamente:

````python
def autores(redeSocial):
    todosAutores = [post.get('autor') for post in redeSocial if post.get('autor') is not None] #Usar List Comprehension para sacar todos os autores (usando .get() por segurança)
    autoresUnicos = list(set(todosAutores))         #Usar set() para apagar os repetidos e transformar de volta numa lista limpa
    autoresUnicos.sort()
    return autoresUnicos
````

#### d) `insPost`, que acrescenta um novo post à rede social a partir dos parâmetros recebidos e devolve a nova rede social. 
    
O campo `id` devrá ser calculado a partir dos já existentes, por exemplo, se a rede tiver posts com id `p1`, `p2` e `p3`, o novo `id` deverá ser `p4`.

````python
def insPost(redeSocial, conteudo, autor, dataCriacao, comentarios):
    novoNumero = len(redeSocial) + 1    #calcular o novo id, se a lista tem x elementos, ele le esses 3 e o novo post será x + 3. 
    novoId = 'p' + str(novoNumero)      #o comando str converte o nº para texto para podermos colocar o p
    
    novoPost = {        #cria o dicionário do novo post com as chaves certas
        'id': novoId,
        'conteudo': conteudo,
        'autor': autor,
        'dataCriacao': dataCriacao,
        'comentarios': comentarios
    }

    redeSocial.append(novoPost) #adiciona o novo post ao fim do dicionário original
    
    return redeSocial
````

#### e)  `remPost`, que remove um post da rede, correspondente ao `id` recebido.

````python
def remPost(redeSocial, id):
    redeSocial = [post for post in redeSocial if post.get('id') != id]# Mantém na rede social apenas os posts cujo 'id' seja DIFERENTE do id recebido
    return redeSocial
````

#### f) `postsPorAutor`, que devolve uma distribuição de posts por autor (à semelhança do que foi feito nas aulas).

````python
def postsPorAutor(redeSocial):
    distribuicao = {}
    
    for post in redeSocial:
        autor = post.get('autor')# Tenta descobrir quem é o autor deste post 
        if autor in distribuicao:
            distribuicao[autor] = distribuicao[autor] + 1# Se o autor já está no nosso dicionário de contagem, soma mais 1 post
        else:
            distribuicao[autor] = 1# Se for a primeira vez que vemos este autor, o contador começa no 1
            
    return distribuicao
````

#### g) `comentadoPor`, que recebe um autor e devolve a lista de posts comentados por esse autor.

````python
def comentadoPor(redeSocial, autor):
    postsEncontrados = [] # Criamos a lista vazia para guardar os posts finais

    for post in redeSocial: # Inicia o ciclo para analisar cada post da rede social
        listaComentarios = post.get('comentarios', []) # Extrai a lista de comentários (lista vazia se não houver)

        for comentario in listaComentarios: # Inicia o ciclo para analisar cada comentário do post

            if comentario.get('autor') == autor: # Verifica se o autor do comentário é o que procuramos

                if post not in postsEncontrados: # Garante que o post só é guardado se não for repetido
                    postsEncontrados.append(post) # Adiciona o post à lista
                    
    return postsEncontrados # Devolve a lista completa com os posts encontrados
````
