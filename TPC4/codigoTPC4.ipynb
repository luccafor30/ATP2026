```python
listaInterna = []
import random
entrada = input("Você deseja entrar no menu do programa?")

while entrada != "sim" and entrada != "Sim" and entrada != "Não" and entrada != "não" and entrada != "nao" and entrada != "Nao":
    print("Não entendi, responda apenas sim ou não.")
    entrada = input("Você deseja entrar no menu do programa?")
if entrada == "nao" or entrada == "não":
    print("Entendido, a aplicação foi encerrada. Até à próxima!")
elif entrada == "Sim" or entrada == "sim":
    opcao = "-1"
    while opcao != "0":
        print("     Aplicação para manipulação de números inteiros      ")
        print("(1) Criar Lista")
        print("(2) Ler Lista")
        print("(3) Soma")
        print("(4) Média")
        print("(5) Maior")
        print("(6) Menor")
        print("(7) estaOrdenada por ordem crescente")
        print("(8) estaOrdenada por ordem decrescente")
        print("(9) Procura um elemento")
        print("(0) Sair")
        opcao = input("Escolha uma opção").strip()
        
        if opcao == "1":
            print("Escolheu criar uma lista aleatória")
            listaInterna=[]
            n = int(input("Quantos números você quer que a sua lista tenha?"))
            for i in range(n):
                numero = random.randint(1,100)
                listaInterna.append(numero)
            print(listaInterna)


        elif opcao == "2":
            print("Escolheu ler uma lista")
            listaInterna=[]
            n = int(input("Quantos números você quer que a sua lista tenha?"))
            for i in range(n):
                numero = int(input("Escolha um número"))
                listaInterna.append(numero)
            print(listaInterna)

            
        elif opcao == "3":
            print("Escolheu calcular a Soma")
            if len(listaInterna) > 0:
                print(f"A soma dos números da lista é {sum(listaInterna)}")
            else:
                print("A lista está vazia! Por favor, crie ou leia uma lista primeiro nas opções 1 ou 2.")
            print(listaInterna)
            
        elif opcao == "4":
            print("Escolheu calcular a Média")
            if len(listaInterna) > 0:
                Media = sum(listaInterna) / len(listaInterna)
                print(f"A média dos números da lista é: {Media}")
            else:
                print("A lista está vazia! Por favor, crie ou leia uma lista primeiro nas opções 1 ou 2.")
            print(listaInterna)

        elif opcao == "5":
            print("Escolheu descobrir o Maior")
            if len(listaInterna) > 0:
                maior = listaInterna[0]
                for x in listaInterna:
                    if maior<x:
                        maior = x
                print(f"O Maior número é {maior}")
            else:
                print("A lista está vazia! Por favor, crie ou leia uma lista primeiro nas opções 1 ou 2.")
            print(listaInterna)

        elif opcao == "6":
            print("Escolheu descobrir o Menor")
            if len(listaInterna) > 0:
                menor = listaInterna[0]
                for x in listaInterna:
                    if menor>x:
                        menor = x
                print(f"O Menor número é {menor}")
            else:
                print("A lista está vazia! Por favor, crie ou leia uma lista primeiro nas opções 1 ou 2.")
            print(listaInterna)

        elif opcao == "7":
            print("Escolheu verificar Ordem Crescente")
            if len(listaInterna) > 0:
                i = 0
                res = True
                while i < len(listaInterna) - 1 and res:
                    if listaInterna[i] > listaInterna[i + 1]:
                        res = False
                    i = i + 1
                if res==True:
                    print(f"A lista está ordenada por ordem crescente. A lista é {listaInterna}")
                else:
                    print(f"A lista não está ordenada por ordem crescente, A lista é {listaInterna}")
            else:
                print("A lista está vazia! Por favor, crie ou leia uma lista primeiro nas opções 1 ou 2.")
            
        elif opcao == "8":
            print("Escolheu verificar Ordem Decrescente")
            if len(listaInterna) > 0:
                i = 0
                res = True
                while i < len(listaInterna) - 1 and res:
                    if listaInterna[i] < listaInterna[i + 1]:
                        res = False
                    i = i + 1
                if res==True:
                    print(f"A lista está ordenada por ordem decrescente. A lista é {listaInterna}")
                else:
                    print(f"A lista não está ordenada por ordem decrescente, A lista é {listaInterna}")
            else:
                print("A lista está vazia! Por favor, crie ou leia uma lista primeiro nas opções 1 ou 2.")
            
        elif opcao == "9":
            print("Escolheu Procurar um elemento ")
            if len(listaInterna) > 0:
                alvo=int(input("Escolha o número que quer encontrar e eu te digo a posição"))
                i=0
                posicao=-1
                encontrei=False
                while i<len(listaInterna) and encontrei==False:
                    if listaInterna[i]==alvo:
                        posicao=i
                        encontrei=True
                    i = i + 1
                if encontrei==True:
                    print(f"O número que pediu se encontra na posição de indice {posicao}")
                else:
                    print(f"Nessa lista não há o número pedido")
            else:
                print("A lista está vazia! Por favor, crie ou leia uma lista primeiro nas opções 1 ou 2.")
            print(listaInterna)
            
        elif opcao == "0":
            print(f"A encerrar a aplicação. A tua lista final guardada é: {listaInterna}")
            
        else:
            print("Opção inválida! Escolhe um número de 0 a 9.")
```
