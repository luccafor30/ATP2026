## 100 Lâmpadas

![o corredor das 100 lâmpadas](lampadas.jpeg)

Num corredor existem 100 lâmpadas e os respetivos 100 interruptores, inicialmente com todas as luzes apagadas. 
Uma pessoa percorre esse corredor 100 vezes, seguindo uma regra específica para alterar o estado das lâmpadas (se está apagada, acende; se está acesa, apaga):

- Na 1ª passagem, altera todos os interruptores (1, 2, 3, 4...).
- Na 2ª passagem, altera os interruptores de 2 em 2 (2, 4, 6, 8...).
- Na 3ª passagem, altera os interruptores de 3 em 3 (3, 6, 9, 12...).
- O padrão continua sucessivamente até à 100ª passagem, em que altera apenas o 100º interruptor.

**Problema**: No final de todas as 100 passagens, quais são as lâmpadas que terminam acesas?

### Aquecimento neuronal/algorítmico

Dada uma lista de números inteiros e um valor alvo, escreva um programa em Python para encontrar todos os pares de elementos na lista cuja soma seja igual ao valor alvo. Você deve retornar uma lista de pares de elementos que satisfaçam essa condição.

Por exemplo: 

In > `[2, 7, 4, -1, 5, 9]` e `6` 

Out > `[(2, 4), (-1, 7)]`

```python
def procuraSoma(lista, valor):
    res = []
    i = 1
    for x in lista:
        for y in lista[1:]:
            if x + y == valor:
                res.append((x,y))
        i = i + 1
    return res                                  

print(procuraSoma([2,7,4,-1,5,9,3], 6))
```

```python
#outra forma com index
def procuraSoma(lista, valor):
    res = []
    for x in lista:
        for y in lista[lista.index(x)+1]:
            if x + y == valor:
                res.append((x,y))
        i = i + 1
    return res                                  

print(procuraSoma([2,7,4,-1,5,9,3], 6))
```

### Abertura: open(nome, modo)

Modos de abertura de um ficheiro:
* "r" - Read - Valor por omissão, siginifca q se n disser nada este é oq devemos contar. Abre o ficheiro para leitura, devolve error se o ficheiro não existir;
* "a" - Append - Abre o ficheiro para acrescentar, cria o ficheiro se este não existir;
* "w" - Write - Abre o ficheiro para escrever, cria o ficheiro se este não existir, apaga o conteúdo se este existir;
* "x" - Create - Cria o ficheiro, devolve error se o ficheiro já existir.

Parâmetro adicional ao modo:
* "t" - Text - Valor por omissão, siginifca q se n disser nada este é oq devemos contar. Ficheiro textual;
* "b" - Binary - Ficheiro binário, por exemplo, imagens.

### Resumo

A seguir apresenta-se um resumo das funcionalidades sobre dicionários.

#### Operadores

* `expressão in dicionário` - verifica se o valor da expressão é uma das chaves ("keys") do cicionário retornando `True` ou `False`;
* `expressão not in dicionário` - verifica se o valor da expressão não é uma das chaves ("keys") do cicionário retornando `True` ou `False`;
* `len(dicionário)` - retorna o número de pares chave-valor no dicionário;

#### Métodos (subjacentes ao tipo de dados)

* `d.clear()` - apaga o conteúdo de um dicionário ficando este vazio: `{}`;
* `d.get(key, [default])` - retorna o valor associado a uma chave sem dar origem a um erro, se a chave não existir retorna a constante `None`, se `default` for especificado é esse valor que é retornado em lugar da constante;
* `d.items()` - retorna uma lista de tuplos (chave, valor) de todas as entradas no dicionário;
* `d.keys()` - retorna a lista de chaves no dicionário;
* `d.values()` - retorna a lista de valores no dicionário;
* `d.pop(key, [default])` - remove a entrada associada à chave passada como argumento do dicionário e retorna o respetivo valor, se a chave não existir dá erro a não ser que o valor por omissão seja especificado, nesse caso, é esse o valor retornado;
* `d.popitem()` - remove o último par chave-valor adicionado ao dicionário, se o dicionário estiver vazio gera um erro;
* `d.update(obj)` - funde o dicionário com o objeto, podendo este ser outro dicionário ou qualquer estrutura que tenha pares;
