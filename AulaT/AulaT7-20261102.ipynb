## O Enigma dos 5 Piratas e o Tesouro

Cinco piratas (ordenados por hierarquia: do Capitão A, o mais alto, ao Pirata E, o mais baixo) precisam de dividir um tesouro de 100 moedas de ouro.

A regra de divisão é a seguinte:
- O pirata de maior hierarquia ainda vivo propõe uma forma exata de dividir o ouro;
- Todos os piratas vivos (incluindo o que fez a proposta) votam a favor ou contra;
- Se a proposta receber 50% ou mais dos votos, é aprovada e o tesouro é dividido;
- Se for rejeitada, o pirata que fez a proposta é atirado ao mar e o próximo pirata na hierarquia faz uma nova proposta, repetindo-se o processo.

Os piratas obedecem a três regras absolutas de comportamento: 
- são perfeitamente lógicos; 
- são extremamente gananciosos (querem o máximo de moedas possível); 
- e são cruéis (se a quantidade de moedas que receberem for igual em dois cenários, preferem votar para atirar o líder ao mar).

### Que divisão exata deve o Capitão A propor para sobreviver e reter a maior quantidade de moedas possível?

---

### 💡 Ligação à Recursividade: Indução Retroativa (*Backward Induction*)

Este problema ilustra o modelo mental da recursividade: em vez de tentar adivinhar a partir dos 5 piratas em simultâneo, resolve-se identificando o **caso base** (a situação mais simples possível) e construindo indutivamente a solução para os casos de maior dimensão:

1. **Caso Base (se apenas sobrarem 2 piratas: D e E)**:  
   - D propõe: 100 para D, 0 para E.  
   - D vota a favor (1 em 2 = 50%), a proposta passa. E recebe 0 moedas.

2. **Passo 1 (se sobrarem 3 piratas: C, D e E)**:  
   - C sabe que se morrer, avança-se para o caso de 2 piratas onde E recebe 0.  
   - Logo, basta a C oferecer 1 moeda a E para comprar o seu voto (1 > 0).  
   - Proposta de C: `C: 99, D: 0, E: 1`. Votos de C e E (2 em 3 = 66%) -> Aprovada!

3. **Passo 2 (se sobrarem 4 piratas: B, C, D e E)**:  
   - B precisa de 50% dos votos (2 em 4: o seu e mais 1).  
   - Se B morrer, o cenário de C daria 0 moedas a D.  
   - Logo, B compra o voto de D com apenas 1 moeda.  
   - Proposta de B: `B: 99, C: 0, D: 1, E: 0`. Votos de B e D (50%) -> Aprovada!

4. **Passo 3 (os 5 piratas: A, B, C, D e E)**:  
   - O Capitão A precisa de 50% dos votos (3 em 5: o seu e mais 2).  
   - No cenário de B, os piratas C e E recebem 0 moedas.  
   - A compra os votos de C e E oferecendo 1 moeda a cada um.  
   - **Proposta ótima de A**: `A: 98, B: 0, C: 1, D: 0, E: 1`.  
   - Votam a favor: A, C e E (3 em 5 = 60%). Aprovada!

### Aquecimento neuronal/algorítmico

Especifica uma função que dado um parâmetro `n` gera os primeiros `n` números naturais da sequência de Fibonacci e os devolve numa lista.

Os dois primeiros elementos desta sequência são `1`, os restantes obtêm-se pela soma dos dois anteriores.

Por exemplo: 

In > `9` 

Out > `[1, 1, 2, 3, 5, 8, 13, 21, 34]`

*Nota: Lembra-te de tratar casos de fronteira como n <= 0 (lista vazia) e n == 1 ([1]).*

````python
def aquecimento(n):
    res = []
    if n <= 0:
        return res
    elif n == 1:
        return [1]
    
    res = [1, 1]
    while len(res) < n:
        res.append(res[-1] + res[-2])
    return res

for i in range(1, 20):
    print(f"Fibonacci({i:2d}) -> {aquecimento(i)}")
````

## Recursividade

Uma função diz-se **recursiva** quando é definida em termos de si própria (invoca-se a si mesma).

Toda a função recursiva bem desenhada necessita de:
1. **Caso Base (Condição de Paragem)**: A instância mais simples do problema, calculada diretamente sem nova chamada recursiva;
2. **Passo Recursivo**: Invocação da função com argumentos estritamente menores/mais simples, aproximando a computação do caso base.

```
Visualização da Pilha de Chamadas (Call Stack) para fact(3):
----------------------------------------------------------
fact(3) = 3 * fact(2)
               |-> 2 * fact(1)
                        |-> 1            [Caso Base]
               |<- 2 * 1 = 2             [Desempilhamento]
fact(3) = 3 * 2 = 6                      [Resultado Final]
```

### 1. Fatorial

Definição formal indutiva:
$$
\text{fact}(n) = 
\begin{cases} 
1 & \text{se } n \le 1 \quad \text{(Caso Base)} \\
n \times \text{fact}(n - 1) & \text{se } n > 1 \quad \text{(Passo Recursivo)}
\end{cases}
$$

````python
def fact(n):
    """Calcula o fatorial de n recursivamente."""
    if n <= 1:
        return 1
    return n * fact(n - 1)
    
for i in range(0, 11):
    print(f"{i:2d}! = {fact(i)}")       # O 2d serve pra por os iguais alinhados e tbm quantos digitos totais tens, neste caso dezena
````

````python
def factIter(n):
    """Calcula o fatorial de n iterativamente."""
    res = 1
    for val in range(2, n + 1):
        res *= val
    return res

for i in range(0, 11):
    print(f"{i:2d}! = {factIter(i)}")
````

````python
def factIter(n):
    """Calcula o fatorial de n iterativamente."""
    res = 1
    while n >1:
        res = n * res
        n = n - 1
    return res

for i in range(0, 11):
    print(f"{i:2d}! = {factIter(i)}")
````

### 2. Multiplicação baseada na soma

A multiplicação de dois naturais $a \times b$ pode ser definida como a soma repetida de $a$ durante $b$ vezes:
$$
a \times b = 
\begin{cases} 
0 & \text{se } b = 0 \quad \text{(Caso Base)} \\
a + (a \times (b - 1)) & \text{se } b > 0 \quad \text{(Passo Recursivo)}
\end{cases}
$$

````python
def multiplica(a, b):
    """Calcula a * b recursivamente por somas sucessivas."""
    if b == 0:
        return 0
    else:
        return a + multiplica(a, b - 1)
    
print("8 * 9   =", multiplica(8, 9))
print("11 * 14 =", multiplica(11, 14))
print("7 * 0   =", multiplica(7, 0))
````

````python
def multiplicaIter(a, b):
    """Calcula a * b iterativamente."""
    res = 0
    while b != 0:
        res = a + res           #repete uma soma constante, por exemplo se temos 8 vezes 9, teremos o 8 a somar-se 9 vezes, onde o b é o 9, e o ciclo se repete até 9 chegar a 0, de -1 em -1 a cada vez, e o resultado é a soma de 8 ao res 9 vezes
        b = b - 1
    return res

print("8 * 9   =", multiplicaIter(8, 9))
print("11 * 14 =", multiplicaIter(11, 14))
print("7 * 0   =", multiplicaIter(7, 0))
````

````python
def multiplicaIter(a, b):
    """Calcula a * b iterativamente."""
    res = 0
    for _ in range(b):
        res += a
    return res

print("8 * 9   =", multiplicaIter(8, 9))
print("11 * 14 =", multiplicaIter(11, 14))
print("7 * 0   =", multiplicaIter(7, 0))
````

### 3. Comprimento de uma lista

Raciocínio indutivo sobre sequências:
- **Caso Base**: O comprimento da lista vazia `[]` é `0`;
- **Passo Recursivo**: O comprimento de uma lista não-vazia é `1 + compLista(resto da lista)`.

````python
def compLista(lista):
    """Calcula o comprimento de uma lista recursivamente."""
    if lista == []:
        return 0
    return 1 + compLista(lista[1:])
    
def compListaIter(lista):
    """Calcula o comprimento de uma lista iterativamente."""
    res = 0
    while lista != []:
        res = 1 + res
        lista = lista[1:]
    return res
````

````python
lista = "Quinta feira"
print("Fatia a partir de 1:", lista[1:])
print("Comprimento (Iterativo):", compListaIter(list(lista)))
print("Comprimento (Recursivo):", compLista(list(lista)))
````

### Carregar um dataset JSON

Utilizamos a construção `with open(...)` (context manager) que garante o encerramento seguro do ficheiro.

````python
import json

ficheiro_emd = "./ficheiros/emd.json"

with open(ficheiro_emd, "r", encoding="utf-8") as f:
    emds = json.load(f)

print("Tipo do dataset carregado:", type(emds))
print(f"Total de registos: {len(emds)}")
print("Primeiro registo:", emds[0])
print("Nome da pessoa no primeiro registro:", emds[0]["nome"]["primeiro"], emds[0]["nome"]["ultimo"])  #achar apenas os nomes do dicionário
````

#### Escrever no monitor: nome do atleta, modalidade, clube (ordenada alfabeticamente por nome)

### Perguntas:
1. **Como devo guardar uma string JSON num ficheiro?**  
   Se tiveres uma string JSON (`s = json.dumps(...)`), podes escrever com `fout.write(s)`. Contudo, é muito mais eficiente passar o objeto Python diretamente a `json.dump(obj, fout)`.
2. **Quando devo usar o `json.dump()`?**  
   Quando queres gravar o objeto Python diretamente num ficheiro aberto em disco. Usa-se `json.dumps()` quando queres o resultado numa string Python em memória.

````python
# Escrever a informação num ficheiro JSON com indentação
with open('./ficheiros/myemd.json', 'w', encoding='utf-8') as fout:
    json.dump(emds, fout, indent=4, ensure_ascii=False)

print("Ficheiro ./ficheiros/myemd.json gravado com sucesso!")
````

#### Problemas a fazer na aula:

1. Qual a distribuição por **género** dos atletas que fizeram EMD?
2. Qual a distribuição por **modalidade** dos atletas que fizeram EMD?
3. Especifica uma função para fazer um **gráfico de barras** de uma distribuição e aplica-a às distribuições calculadas.
4. Guarda as distribuições em ficheiro JSON.

````python
def distribGen(emdLista):
    """Calcula a distribuição de atletas por género."""
    res = {}
    for atleta in emdLista:
        genero = atleta.get('género', 'Outro')
        res[genero] = res.get(genero, 0) + 1
    return res

distrib_gen = distribGen(emds)
print("Distribuição por Género:", distrib_gen)
````

````python
def distribModalidade(emdLista):
    """Calcula a distribuição de atletas por modalidade."""
    res = {}
    for atleta in emdLista:
        modalidade = atleta.get('modalidade', 'Desconhecida')
        res[modalidade] = res.get(modalidade, 0) + 1
    return res

distribMod = distribModalidade(emds)
distribModLista = sorted(distribMod.items(), key=lambda par: par[1], reverse=True)
print("Distribuição completa por modalidade:", distribModLista)
print("\nTop 3 Modalidades mais procuradas:", distribModLista[:3])
````

#### Gráficos de Barras com `matplotlib`

Função para desenhar e visualizar distribuições estatísticas:

````python
import matplotlib.pyplot as plt

def desenhaGraficoBarras(distrib, titulo="Distribuição", xlabel="Categorias", ylabel="Contagem"):
    """Gera um gráfico de barras a partir de um dicionário {categoria: contagem}."""
    categorias = list(distrib.keys())
    valores = list(distrib.values())
    
    plt.figure(figsize=(10, 5))
    barras = plt.bar(categorias, valores, color="royalblue", edgecolor="black")
    plt.title(titulo, fontsize=14, fontweight="bold")
    plt.xlabel(xlabel, fontsize=12)
    plt.ylabel(ylabel, fontsize=12)
    plt.xticks(rotation=45, ha="right")
    plt.grid(axis="y", linestyle="--", alpha=0.7)
    
    for barra in barras:
        altura = barra.get_height()
        plt.annotate(f"{altura}",
                     xy=(barra.get_x() + barra.get_width() / 2, altura),
                     xytext=(0, 3), textcoords="offset points",
                     ha="center", va="bottom", fontsize=9)
    
    plt.tight_layout()
    plt.show()

# Exemplos de aplicação:
desenhaGraficoBarras(distrib_gen, titulo="Distribuição por Género dos Atletas", xlabel="Género")
desenhaGraficoBarras(dict(distribModLista[:8]), titulo="Top 8 Modalidades Mais Praticadas", xlabel="Modalidade")
````

````python
# Guardar uma distribuição em ficheiro JSON
def distribGuardar(d, fnome):
    """Guarda um dicionário num ficheiro JSON sob a pasta datasets."""
    caminho = f"./ficheiros/{fnome}.json"
    with open(caminho, "w", encoding="utf-8") as fout:
        json.dump(d, fout, indent=4, ensure_ascii=False)
    print(f"Distribuição guardada com sucesso em {caminho}")

distribGuardar(distribMod, "distrib-modalidades")
````

````python
# Recuperar uma distribuição de um ficheiro JSON
def distribRecuperar(fnome):
    """Carrega um dicionário de um ficheiro JSON sob a pasta datasets."""
    caminho = f"./ficheiros/{fnome}.json"
    with open(caminho, "r", encoding="utf-8") as f:
        return json.load(f)

modalidades = distribRecuperar("distrib-modalidades")
print(f"Tipo recuperado: {type(modalidades)}")
print("Primeiras 5 modalidades recuperadas:", list(modalidades.items())[:5])
````

## Operações CRUD

O acrónimo **CRUD** sintetiza as quatro operações essenciais de qualquer sistema de informação:
* **(C) Create**: Inserir/criar um novo registo;
* **(R) Retrieve**: Recuperar, consultar ou listar registos existentes;
* **(U) Update**: Editar ou atualizar a informação de um registo;
* **(D) Delete**: Apagar ou remover um registo.

Abaixo definimos as funções CRUD sobre a lista de EMDs mantida em memória.

````python
# (C) Create: Inserir um EMD novo
def inserirEMD(bd, novo_emd):
    """Insere um novo registo na base de dados, garantindo unicidade do _id."""
    for emd in bd:
        if emd['_id'] == novo_emd['_id']:
            print(f"[Erro] Registo com _id '{novo_emd['_id']}' já existe!")
            return False
    bd.append(novo_emd)
    print(f"[Sucesso] Registo '{novo_emd['_id']}' inserido com sucesso.")
    return True
````

````python
# (R) Retrieve: Consultar um EMD por _id
def consultarEMD(bd, id_registo):
    """Devolve o registo correspondente ao _id ou None se não existir."""
    for emd in bd:
        if emd['_id'] == id_registo:
            return emd
    return None
````

````python
# (R) Retrieve: Listar os EMDs de forma tabular
def listarEMDs(bd, limite=10):
    """Apresenta no ecrã os primeiros N registos da base de dados."""
    print(f"{'ID':<26} | {'Nome':<25} | {'Modalidade':<15} | {'Resultado'}")
    print("-" * 80)
    for emd in bd[:limite]:
        nome = f"{emd['nome']['primeiro']} {emd['nome']['último']}"
        res = "Apto" if emd.get('resultado') else "Não Apto"
        print(f"{emd['_id']:<26} | {nome:<25} | {emd.get('modalidade', ''):<15} | {res}")
````

````python
# (U) Update: Alterar um EMD existente
def alterarEMD(bd, id_registo, novos_dados):
    """Atualiza campos de um registo já existente."""
    emd = consultarEMD(bd, id_registo)
    if emd:
        emd.update(novos_dados)
        print(f"[Sucesso] Registo '{id_registo}' atualizado com sucesso.")
        return True
    print(f"[Erro] Registo '{id_registo}' não encontrado para atualização.")
    return False
````

````python
# (D) Delete: Apagar um EMD por _id
def apagarEMD(bd, id_registo):
    """Remove o registo com o _id indicado da base de dados."""
    for i, emd in enumerate(bd):
        if emd['_id'] == id_registo:
            removido = bd.pop(i)
            print(f"[Sucesso] Registo '{id_registo}' removido com sucesso.")
            return removido
    print(f"[Erro] Registo '{id_registo}' não encontrado para remoção.")
    return None
````

````python
# Demonstração prática do ciclo completo de operações CRUD

# 1. Criar novo registo
novo_atleta = {
    "_id": "emd_demo_001",
    "index": len(emds),
    "dataEMD": "2026-11-02",
    "nome": {"primeiro": "Carlos", "último": "Martins"},
    "idade": 22,
    "género": "M",
    "morada": "Braga",
    "modalidade": "Natação",
    "clube": "SCBraga",
    "email": "carlos.martins@scbraga.pt",
    "federado": True,
    "resultado": True
}

print("--- 1. CREATE ---")
inserirEMD(emds, novo_atleta)

print("\n--- 2. RETRIEVE (Consulta Individual) ---")
print("Consulta:", consultarEMD(emds, "emd_demo_001"))

print("\n--- 3. UPDATE ---")
alterarEMD(emds, "emd_demo_001", {"modalidade": "Triatlo", "idade": 23})
print("Após alteração:", consultarEMD(emds, "emd_demo_001"))

print("\n--- 4. DELETE ---")
apagarEMD(emds, "emd_demo_001")
print("Após remoção:", consultarEMD(emds, "emd_demo_001"))
````
