## O Campeonato de jogos tipo _"Arcade"_

Imagine que um salão de jogos antigo regista as pontuações das máquinas ao longo do dia. 
O sistema exporta os dados de forma desorganizada através de uma lista de tuplos, onde cada tuplo contém o nome do jogador e os pontos ganhos num jogo específico. 
Um mesmo jogador pode ter jogado várias vezes ao longo do dia.

Exemplo:

```python
registos_do_dia = [
    ("Ana", 150), ("Bruno", 200), ("Ana", 50), 
    ("Carlos", 300), ("Bruno", 100), ("Ana", 200), 
    ("Carlos", 50)
]
```

Especifica um programa ou função que calcula a pontuação total de cada jogador para coroar o vencedor do dia.

````python
# Cálculo da pontuação final
registos_do_dia = [
    ("Ana", 150), ("Bruno", 200), ("Ana", 50), 
    ("Carlos", 300), ("Bruno", 100), ("Ana", 200), 
    ("Carlos", 50)
]

def pertence(ListaTuplos, nome):
    encontrado = False
    i = 0
    while not encontrado and i<len(ListaTuplos):
        if nome == ListaTuplos[i][0]:
            encontrado = True
        i = i + 1
    return encontrado

def posicao(ListaTuplos, nome):
    encontrado = False               
    i = 0
    while not encontrado and i<len(ListaTuplos):
        if nome == ListaTuplos[i][0]:
            encontrado = True
        else:
            i = i + 1
    return i

def calcPontFinal(registos):
    res = []
    for nome, pontos in registos:
        if pertence(res, nome):
            res[posicao(res, nome)] = (nome, res[posicao(res, nome)][1] + pontos)
        else:
            res.append((nome, pontos))
    return res

print(calcPontFinal(registos_do_dia))
````

````python
#FORMA MAIS FÁCIL DE FAZER ESSA MERDA
def calcPontFinalDict(registos):
    res = {}
    for nome, pontos in registos:
        if nome in res.keys():
            res[nome] = res[nome] + pontos
        else:
            res[nome] = pontos
    return res

print(calcPontFinalDict(registos_do_dia))
````

## Dicionários (Arrays Associativos, Objetos, Produto Cartesiano)

### Conceitos

- Dicionário: Lista de pares chave --> valor;
- As chaves são únicas (não se repetem) e imutáveis;
- Permitem um acesso direto por chave.

Exemplo:

```python
mealheiro = {
    '10c': 5,
    '20c': 3,
    '50c': 7,
    '1e': 12,
    '2e': 6
}
```

_Quanto dinheiro tenho no mealheiro?_

````python
mealheiro = {
    '10c': 5,
    '20c': 3,
    '50c': 7,
    '1e': 12,
    '2e': 6
}

def calcValor(m):
    res = 0.0
    for chave, valor in m.items():
        if chave[-1] == "c":
            res = res + (int(chave[:-1]) * valor) / 100
        else:
            res = res + (int(chave[:-1]) * valor)
    return res

print(calcValor(mealheiro))
````

### Iteração em dicionários

- Iterar pelas chaves: `for chave in dic:`;
- Iterar pelos valores: `for valor in dic.values():`;
- Iterar por ambos (chaves e valores): `for chave, valor in dic.items():`.

````python
# Exemplos:
turma1 = {
    "A4140": ("José Carlos Ramalho", "LESI"),
    "A4238": ("José Alberto Rodrigues", "LESI"),
    "A4156": ("Paulo Jorge Domingues", "LESI")
}

# 1. Iterar pelas chaves
for chave in turma1:
    print(chave, turma1[chave])     #turma1[chuve] é as informações que têm dentro da chave
````

````python
# 2. Iterar pelos valores
for valor in turma1.values():
    print(valor)
````

````python
# 3. Iterar por ambos (chaves e valores)
for chave,valor in turma1.items():
    print(chave, valor)
````

### Resumo

A seguir apresenta-se um resumo das funcionalidades sobre dicionários.

#### Operadores

* `expressão in dicionário` - verifica se o valor da expressão é uma das chaves ("keys") do cicionário retornando `True` ou `False`;
* `expressão not in dicionário` - verifica se o valor da expressão não é uma das chaves ("keys") do cicionário retornando `True` ou `False`;
* `len(dicionário)` - retorna o número de pares chave-valor no dicionário;

#### Métodos (subjacentes ao tipo de dados)

* `d.clear()` - apaga o conteúdo de um dicionário ficando este vazio: `{}`;
* `d.get(key, [default])` - retorna o valor associado a uma chave sem dar origem a um erro, se a chave não existir retorna a constante `None`, se `default` for especificado é esse valor que é retornado em lugar da constante;
* `d.items()` - retorna uma lista de tuplos (chave, valor) de todas as entradas no dicionário;
* `d.keys()` - retorna a lista de chaves no dicionário;
* `d.values()` - retorna a lista de valores no dicionário;
* `d.pop(key, [default])` - remove a entrada associada à chave passada como argumento do dicionário e retorna o respetivo valor, se a chave não existir dá erro a não ser que o valor por omissão seja especificado, nesse caso, é esse o valor retornado;
* `d.popitem()` - remove o último par chave-valor adicionado ao dicionário, se o dicionário estiver vazio gera um erro;
* `d.update(obj)` - funde o dicionário com o objeto, podendo este ser outro dicionário ou qualquer estrutura que tenha pares;
